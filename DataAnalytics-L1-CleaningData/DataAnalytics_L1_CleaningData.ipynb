```python

import pandas as pd
import numpy as np

df = pd.read_csv('/content/train.csv')

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python

print("DATA QUALITY REPORT")

print("\n1. Dataset Information:")
df.info()

print("\n2. Missing Values:")
print(df.isnull().sum())

print("\n3. Duplicate Rows:")
print(df.duplicated().sum())

print("\n4. Dataset Size:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\n5. Statistical Summary:")
display(df.describe(include="all"))
```

```python

# Fill missing Age values with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Embarked values with mode
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Cabin has many missing values, so drop the column
df = df.drop(columns=['Cabin'])

# Check missing values after handling
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python

# Count duplicate rows before removal
duplicates_before = df.duplicated().sum()
print("Duplicate rows before removal:", duplicates_before)

# Remove duplicate rows
df = df.drop_duplicates()

# Count duplicate rows after removal
print("Duplicate rows after removal:", df.duplicated().sum())

print("Dataset shape after removing duplicates:", df.shape)
```

```python

# Standardise Sex values
df['Sex'] = df['Sex'].str.strip().str.lower().map({
    'male': 'Male',
    'female': 'Female'
})

# Standardise Embarked values
df['Embarked'] = df['Embarked'].str.strip().str.upper()

# Display standardised values
print("Sex values:", df['Sex'].unique())
print("Embarked values:", df['Embarked'].unique())
```

```python

# Detect outliers using IQR method

for col in ['Age', 'Fare']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col} outliers: {len(outliers)}")
    print(f"Lower limit: {lower:.2f}")
    print(f"Upper limit: {upper:.2f}")
    print()
```

```python

# Correct data types
df['PassengerId'] = df['PassengerId'].astype(str)
df['Survived'] = df['Survived'].astype(int)
df['Pclass'] = df['Pclass'].astype(int)
df['Age'] = df['Age'].astype(float)
df['Fare'] = df['Fare'].astype(float)

# Display updated data types
print("Updated Data Types:")
print(df.dtypes)
```

```python

# Before and After Cleaning Summary

summary = pd.DataFrame({
    'Metric': [
        'Total Rows',
        'Total Columns',
        'Missing Values',
        'Duplicate Rows',
        'Data Type Accuracy'
    ],
    'Before Cleaning': [
        891,
        12,
        866,
        0,
        'Needs correction'
    ],
    'After Cleaning': [
        len(df),
        len(df.columns),
        int(df.isnull().sum().sum()),
        int(df.duplicated().sum()),
        'Corrected'
    ]
})

display(summary)
```

```python

# Save cleaned dataset
df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
print("Final dataset shape:", df.shape)
```
